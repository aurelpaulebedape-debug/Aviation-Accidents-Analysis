## Data Loading and Inspection

## Data Cleaning

### Filtering aircrafts and events

We want to filter the dataset to include aircraft that the client is interested in an analysis of:
- inspect relevant columns
- figure out any reasonable imputations
- filter the dataset

```python
columns = [
    "Event.Date",
    "Investigation.Type",
    "Aircraft.Category",
    "Amateur.Built",
    "FAR.Description"
]

for col in columns:
    print("\n", col)
    print(df[col].value_counts(dropna=False))

# Convert event dates to datetime
df["Event.Date"] = pd.to_datetime(df["Event.Date"], errors="coerce")

# Keep accidents from 1983 onwards
df = df[
    (df["Event.Date"].dt.year >= 1983) &
    (df["Investigation.Type"] == "Accident") &
    (df["Amateur.Built"] == "No")
].copy()

print(df.shape)
print(df["Aircraft.Category"].value_counts(dropna=False))

missing_category = df[df["Aircraft.Category"].isna()]

print(missing_category["Make"].value_counts().head(20))

print(
    missing_category.groupby(["Make", "Model"])
    .size()
    .sort_values(ascending=False)
    .head(20)
)

known_category = df[df["Aircraft.Category"].notna()]

print(
    known_category.groupby(["Make", "Model", "Aircraft.Category"])
    .size()
    .sort_values(ascending=False)
    .head(25)
)

# Standardize manufacturer and model names
df["Make"] = df["Make"].str.strip().str.upper()
df["Model"] = df["Model"].str.strip().str.upper()

# Use known categories for matching make/model combinations
known = df.dropna(subset=["Aircraft.Category"])

category_map = (
    known.groupby(["Make", "Model"])["Aircraft.Category"]
    .agg(lambda x: x.mode().iloc[0])
)

# Fill missing categories when a matching make/model exists
def fill_category(row):
    if pd.isna(row["Aircraft.Category"]):
        return category_map.get((row["Make"], row["Model"]), np.nan)
    return row["Aircraft.Category"]

df["Aircraft.Category"] = df.apply(fill_category, axis=1)

print(df["Aircraft.Category"].value_counts(dropna=False))

# Keep only airplanes
df = df[df["Aircraft.Category"] == "Airplane"].copy()

print(df.shape)
```

### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
injury_cols = [
    "Total.Fatal.Injuries",
    "Total.Serious.Injuries",
    "Total.Minor.Injuries",
    "Total.Uninjured"
]

print(df[injury_cols].isna().sum())

display(df[injury_cols].describe())

# Replace missing injury counts with zero
df[injury_cols] = df[injury_cols].fillna(0)

# Estimate total persons involved
df["Total.Persons"] = df[injury_cols].sum(axis=1)

# Calculate fatal/serious injury rate
df["Injury.Rate"] = np.where(
    df["Total.Persons"] > 0,
    (df["Total.Fatal.Injuries"] + df["Total.Serious.Injuries"]) / df["Total.Persons"],
    np.nan
)

display(df[["Total.Persons", "Injury.Rate"]].describe())
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python
print(df["Aircraft.damage"].value_counts(dropna=False))

# Create binary indicator for aircraft destruction
df["Aircraft.Destroyed"] = df["Aircraft.damage"].map({
    "Destroyed": 1,
    "Substantial": 0,
    "Minor": 0
})

# Inspect the new column
print(df["Aircraft.Destroyed"].value_counts(dropna=False))
```

### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)

```python
print("Missing values:", df["Make"].isna().sum())
print("Unique makes:", df["Make"].nunique())

make_counts = df["Make"].value_counts()

print(make_counts.head(20))
print("Makes with at least 50 records:", (make_counts >= 50).sum())

# Remove missing manufacturer names
df = df.dropna(subset=["Make"]).copy()

# Keep manufacturers with at least 50 records
make_counts = df["Make"].value_counts()
valid_makes = make_counts[make_counts >= 50].index

df = df[df["Make"].isin(valid_makes)].copy()

print(df.shape)
print(df["Make"].nunique())
```

**Make column cleaning:**
- Standardized manufacturer names by removing extra spaces and converting text to uppercase.
- Removed records with missing manufacturer names.
- Retained manufacturers with at least 50 observations to ensure sufficient data for analysis.
- Some manufacturer naming variations may remain and require further standardization.

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python
# Inspect missing values and model counts
print("Missing models:", df["Model"].isna().sum())
print("Unique models:", df["Model"].nunique())

print(df.groupby(["Make", "Model"]).size().sort_values(ascending=False).head(20))

# Remove rows with missing models
df = df.dropna(subset=["Model"]).copy()

# Check whether model names are shared by different manufacturers
shared_models = df.groupby("Model")["Make"].nunique()

print("Models shared by multiple makes:", (shared_models > 1).sum())

# Create a unique aircraft type identifier
df["Aircraft.Type"] = df["Make"] + " " + df["Model"]

print(df.shape)
print(df["Aircraft.Type"].nunique())
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition
- Number.of.Engines
- Purpose.of.flight
- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python
other_cols = [
    "Engine.Type",
    "Weather.Condition",
    "Number.of.Engines",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]

for col in other_cols:
    print("\n", col)
    print(df[col].value_counts(dropna=False))

# Standardize unknown categories
df["Engine.Type"] = df["Engine.Type"].replace({
    "UNK": "Unknown"
})

df["Weather.Condition"] = df["Weather.Condition"].replace({
    "UNK": "Unknown",
    "Unk": "Unknown"
})

# Standardize flight purpose labels
df["Purpose.of.flight"] = df["Purpose.of.flight"].replace({
    "Air Race show": "Air Race/show"
})

# Verify corrections
print(df["Engine.Type"].value_counts(dropna=False))
print(df["Weather.Condition"].value_counts(dropna=False))
print(df["Purpose.of.flight"].value_counts().loc["Air Race/show"])
```

### Column Removal
- inspect the dataframe and drop any columns that have too many NaNs

```python
# Calculate percentage of missing values per column
missing_pct = (df.isna().sum() / len(df)) * 100

# Display columns sorted by missing percentage
print(missing_pct.sort_values(ascending=False))

# Identify columns with more than 60% missing values
cols_to_drop = missing_pct[missing_pct > 60].index

# Remove these columns
df = df.drop(columns=cols_to_drop)

# Verify the result
print(df.shape)
print(cols_to_drop.tolist())
```
